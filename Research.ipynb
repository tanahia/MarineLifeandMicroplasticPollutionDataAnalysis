### Phase 2: Data Understanding

This research project use two separete datasets to investigate relationship between them.

**Datasets' origins:**
The name of **the first dataset** is Marine Microplastics(Exploring Temporal Trends in Marine Microplastic Density). The dataset was created and uploaded to Kaggle.com by William Zebrowski two years ago.

The records of dataset contain information on microplastics concentrations, date of collection, latitude and longitude where data was collected from 2000 to 2022. The dataset was downloaded on 06/11/2026.

**The second dataset** doesn't have specific name because I created it. I used records from Global Biodiversity Information Facility website, filtered them to match my objective better and downloaded them as csv file. I called this dataset as Marine Species dataset.

The Marine Species dataset containsoccurrence records covering marine species recorded in the Mediterranean and Black Sea regions between 2000 and 2022. It includes their taxonomic information, scientific names, geographical coordinates and observation dates. The dataset was downloaded on 06/11/2026.

**Additionally,** during the initial exploration of the marine species dataset, it was identified that the dataset contains records of both marine and non-marine species.

Since this research focuses specifically on marine species and their potential exposure to microplastic pollution in the Mediterranean and Black Seas, non-marine species records are not relevant to the research objectives.

Therefore, these records will be excluded during the data preparation phase to ensure that the dataset is appropriate for the intended analysis.
